```python
def viz_y_id(y_id):
    max_id = int(y_id.max())
    n_lines = max_id + 1 

    colors = [(0, 0, 0)] + [plt.cm.tab20(i % 20)[:3] for i in range(n_lines)]
    cmap = mcolors.ListedColormap(colors)

    bounds_cmap = [-1.5] + [i - 0.5 for i in range(n_lines + 1)]
    norm = mcolors.BoundaryNorm(bounds_cmap, cmap.N)

    plt.imshow(y_id, cmap=cmap, norm=norm, interpolation='nearest')
```

```python
from typing import Optional

def load_X_with_cloud_threshold(
        identifier:int,
        manager: HedgementationIOManager,
        threshold:float=0.2,
        band_ind:int=0
):
    X,dates = manager.load_X(identifier=identifier, file_type="tif", return_dates=True)
    X_cloud = manager.load_X_cloud(identifier=identifier, file_type="tif")

    X, X_cloud, dates = limit_above_cloud_threshold(X=X, X_cloud=X_cloud, dates=dates, threshold=threshold, band_ind=band_ind)

    return X, X_cloud, dates


def limit_above_cloud_threshold(
    X: np.ndarray,
    X_cloud: np.ndarray,
    dates: Optional[pd.Series]=None,
    threshold:float = 0.2,
    band_ind=0
):
    mean_score_per_timestep = get_mean_cloud_scores(X_cloud, band_ind)

    inds = mean_score_per_timestep > threshold

    X_unobstructed = X[inds, :, :, :]
    cloud_unobstructed = X_cloud[inds, :, :, :]
    if dates is not None:
        dates_unobstructed = dates[inds]
        return X_unobstructed, cloud_unobstructed, dates_unobstructed
    else:
        return X_unobstructed, cloud_unobstructed
```
